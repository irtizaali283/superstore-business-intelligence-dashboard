# Identifies unusually high or low numerical values using the IQR method.

```python
numeric_cols = ["sales", "quantity", "discount", "profit"]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    print(f"{col}: {outliers} potential outliers")
```

# Uses the most frequent Product Name and Product ID matches to create a standardized product ID.

```python
name_id_freq = (
    df.groupby(["product_name", "product_id"])
      .size()
      .reset_index(name="frequency")
)

multi_id_names = (
    df.groupby("product_name")["product_id"]
      .nunique()
)

multi_id_names = multi_id_names[multi_id_names > 1].index

problem_pairs = name_id_freq[
    name_id_freq["product_name"].isin(multi_id_names)
].copy()

problem_pairs["max_frequency"] = (
    problem_pairs.groupby("product_name")["frequency"]
                 .transform("max")
)

top_matches = problem_pairs[
    problem_pairs["frequency"] == problem_pairs["max_frequency"]
]

winner_count = (
    top_matches.groupby("product_name")["product_id"]
               .nunique()
)

resolved_names = winner_count[winner_count == 1].index
tied_names = winner_count[winner_count > 1].index

best_id_map = (
    top_matches[top_matches["product_name"].isin(resolved_names)]
    .set_index("product_name")["product_id"]
    .to_dict()
)

df["standard_product_id"] = df["product_id"]

mask = df["product_name"].isin(resolved_names)

df.loc[mask, "standard_product_id"] = (
    df.loc[mask, "product_name"].map(best_id_map)
)

print("Product names with multiple IDs:", len(multi_id_names))
print("Successfully standardized:", len(resolved_names))
print("Unresolved ties:", len(tied_names))
```

# Resolves equal-frequency ID cases using a fixed rule and checks that the mappings are standardized.

```python
tie_id_map = (
    top_matches[top_matches["product_name"].isin(tied_names)]
    .sort_values(["product_name", "product_id"])
    .drop_duplicates("product_name")
    .set_index("product_name")["product_id"]
    .to_dict()
)

# Apply chosen IDs to tied products
tie_mask = df["product_name"].isin(tied_names)

df.loc[tie_mask, "standard_product_id"] = (
    df.loc[tie_mask, "product_name"].map(tie_id_map)
)

final_check = (
    df[df["product_name"].isin(multi_id_names)]
    .groupby("product_name")["standard_product_id"]
    .nunique()
)

print("Product names with multiple IDs:", len(multi_id_names))
print("Successfully standardized:", (final_check == 1).sum())
print("Still unresolved:", (final_check > 1).sum())
```

# Creates unique standardized IDs for cases where one original ID belongs to different product names.

```python
problem_ids = (
    df.groupby("product_id")["product_name"]
      .nunique()
)

problem_ids = problem_ids[problem_ids > 1].index

mask = df["product_id"].isin(problem_ids)

df.loc[mask, "standard_product_id"] = (
    df.loc[mask, "product_id"]
    + " | "
    + df.loc[mask, "product_name"]
)
```

# Creates useful business features such as year, month, shipping time, profit margin, sales per unit, profit status, and discount groups.

```python
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["month_name"] = df["order_date"].dt.month_name()
df["order_quarter"] = df["order_date"].dt.quarter
df["year_month"] = df["order_date"].dt.to_period("M").astype(str)
df["shipping_days"] = (df["ship_date"] - df["order_date"]).dt.days
df["profit_margin_pct"] = (df["profit"] / df["sales"]) * 100
df["sales_per_unit"] = df["sales"] / df["quantity"]
df["profit_status"] = np.where(
    df["profit"] >= 0,
    "Profit",
    "Loss"
)
df["discount_band"] = pd.cut(
    df["discount"],
    bins=[-0.01, 0, 0.2, 0.5, 1.0],
    labels=["No Discount", "Low", "Medium", "High"]
)
```
